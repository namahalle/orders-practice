```python
from pyspark.sql.functions import col, sum as spark_sum

df_silver = spark.table(
    "practice_rg.default.silver_sales"
)

df_with_revenue = df_silver.withColumn(
    "revenue",
    col("quantity") * col("price")
)

df_gold = (
    df_with_revenue
    .groupBy("city")
    .agg(
        spark_sum("revenue").alias("total_sales")
    )
)

(
    df_gold.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("practice_rg.default.gold_city_sales")
)

display(spark.table("practice_rg.default.gold_city_sales"))
```
