```python
from pyspark.sql.functions import current_timestamp

source_path = "/Volumes/practice_rg/default/volume/sales.csv"

df_raw = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "false")
    .csv(source_path)
)

df_bronze = df_raw.withColumn(
    "ingestion_timestamp",
    current_timestamp()
)

(
    df_bronze.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("practice_rg.default.bronze_sales")
)

display(spark.table("practice_rg.default.bronze_sales"))
```
